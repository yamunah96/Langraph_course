```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
llm= ChatOpenAI(
    model="openai/gpt-oss-120b",
    api_key=os.getenv('GROQ_API_KEY'),
    base_url= "https://api.groq.com/openai/v1",
    max_tokens= 1000
)
```

**Step-1 : Define The schema**

```python
from typing import TypedDict, Literal,List
class graph_schema(TypedDict):
 name:str
 message:str
```

**Step-2 Create the Node functions**

```python
def welcome_message(state:graph_schema) -> graph_schema:
    cuurent_name=  state["name"]
    current_message=  state["message"]
    response=llm.invoke(f"My name is {cuurent_name} :{current_message}").content
    state['message']=f"Your msg was {current_message}.Here's my respones: {response}"
    return state
```

```python
from langgraph.graph import StateGraph,START,END

graph= StateGraph(graph_schema)

# adding nodes
graph.add_node("welcome",welcome_message)
# adding edges
graph.add_edge(START,"welcome")
graph.add_edge("welcome",END)
```

```python
# compile graph
final_graph= graph.compile()
```

```python
from IPython.display import Image, display
display(Image(final_graph.get_graph().draw_mermaid_png()))
```
